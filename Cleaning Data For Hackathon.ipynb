```python
import pandas as pd

# Load the CSV file
df = pd.read_csv('nyc_access.csv')

# View the first 5 rows
print(df.head())
```

```python
# drop population == 0
df = df[df["population"] > 0]
print("after population filter:", len(df))                 

# drop income == 0 or missing
df = df[df["income"].notna() & (df["income"] > 0)]
print("after income filter:", len(df))   

# drop missing poverty_pct
df = df.dropna(subset=["poverty_pct"])
print("after poverty filter:", len(df))                    

# drop every column from B01003_E001 through B01001_M049 (inclusive)
df = df.drop(columns=df.loc[:, "B01003_E001":"B01001_M049"].columns)

df = df.copy()
df.to_csv("nyc_access_clean.csv", index=False)             # new name, original stays untouched
print(df[["population", "income"]].describe())
```
